```python
from sklearn.linear_model import LogisticRegression
import numpy as np
```

```python
x=np.array([[1],[2],[3],[4]])
y=np.array([0,0,1,1])
```

```python
model= LogisticRegression()
model.fit(x,y)
```

```python
print("prediction for 2.5 hours ",model.predict([[2.5]]))
```

```python
print("probability of pass ",model.predict_proba([[2.5]]))
```

```python
#visulize
import matplotlib.pyplot as plt
def sigmoid(x):
  return 1 / (1 +np.exp(-x))
x_vals=np.linspace(-10,10,100)
plt.plot(x_vals,sigmoid(x_vals))
plt.title("sigmoid function")
plt.grid()
plt.show()
```
